## 1. Preparación del entorno

```python
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt

# Si falta alguna librería, activa el entorno virtual .venv de la asignatura e instala:
# python -m pip install -r 03_unidades/U6_automatizacion_comunicacion_resultados/requirements_U6.txt
import pandas as pd
```

## 2. Escribir la función reutilizable

```python
def generar_informe_mensual(ruta_archivo, ruta_salida_excel, ruta_salida_grafico, top_n=10):
    df = pd.read_csv(ruta_archivo)

    resumen_categoria = df.groupby("categoria")["frecuencia"].sum().reset_index()
    resumen_categoria = resumen_categoria.sort_values("frecuencia", ascending=False)

    top_terminos = df.sort_values("frecuencia", ascending=False).head(top_n)

    with pd.ExcelWriter(ruta_salida_excel) as writer:
        resumen_categoria.to_excel(writer, sheet_name="Resumen por categoría", index=False)
        top_terminos.to_excel(writer, sheet_name="Top términos", index=False)

    fig, ax = plt.subplots(figsize=(8, 5))
    ax.barh(top_terminos["termino"][::-1], top_terminos["frecuencia"][::-1])
    ax.set_xlabel("Frecuencia")
    ax.set_title("Términos más frecuentes del mes")
    fig.tight_layout()
    fig.savefig(ruta_salida_grafico)
    plt.close(fig)

    return resumen_categoria, top_terminos
```

## 3. Aplicar la función al informe de enero

```python
resumen_ene, top_ene = generar_informe_mensual(
    "materiales/neologismos_enero.csv",
    "U6_informe_enero.xlsx",
    "U6_grafico_enero.png",
)
resumen_ene
```
