## 2. Importar librerías

```python
from pathlib import Path

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

## 3. Importar datos

```python
ruta_csv = Path("U3_tweets_preparados.csv")
df = pd.read_csv(ruta_csv)
df.head()
```

## 4. Revisar estructura

```python
df.info()
```

```python
df[["longitud_texto", "num_hashtags", "num_menciones", "num_urls"]].describe()
```

## 5. Tablas exploratorias

```python
tabla_fuente = df.groupby("fuente").agg(
    mensajes=("id_str", "count"),
    longitud_media=("longitud_texto", "mean"),
    menciones_medias=("num_menciones", "mean"),
    url_medias=("num_urls", "mean")
).sort_values("mensajes", ascending=False)

tabla_fuente
```

## 6. Distribución de la longitud

```python
plt.figure(figsize=(8, 4))
sns.histplot(data=df, x="longitud_texto", bins=12, color="#2E86AB")
plt.title("Distribución de la longitud de los mensajes")
plt.xlabel("Longitud del texto")
plt.ylabel("Número de mensajes")
plt.tight_layout()
plt.show()
```

## 7. Comparación entre retuits y no retuits

```python
plt.figure(figsize=(7, 4))
sns.boxplot(data=df, x="es_retuit", y="longitud_texto", color="#F4B942")
plt.title("Longitud del texto según marca de retuit")
plt.xlabel("Es retuit")
plt.ylabel("Longitud del texto")
plt.tight_layout()
plt.show()
```

## 8. Relación entre menciones y longitud

```python
plt.figure(figsize=(7, 4))
sns.scatterplot(data=df, x="num_menciones", y="longitud_texto", hue="es_retuit", palette="Set2")
plt.title("Menciones y longitud del texto")
plt.xlabel("Número de menciones")
plt.ylabel("Longitud del texto")
plt.tight_layout()
plt.show()
```

## 9. Heatmap de variables descriptivas

```python
variables = ["longitud_texto", "num_hashtags", "num_menciones", "num_urls"]
correlaciones = df[variables].corr()

plt.figure(figsize=(6, 4))
sns.heatmap(correlaciones, annot=True, cmap="Blues", vmin=-1, vmax=1)
plt.title("Correlación entre variables descriptivas")
plt.tight_layout()
plt.show()
```
