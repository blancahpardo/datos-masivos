## 1. Preparación del entorno

```python
import time
from pathlib import Path

# Si falta pandas, activa el entorno virtual .venv de la asignatura e instala:
# python -m pip install -r 03_unidades/U1_introduccion_analisis_datos_masivos/requirements_U1.txt
import pandas as pd
```

## 2. Cargar los dos catálogos

```python
ruta_pequeno = Path("materiales/U1_catalogo_pequeno.csv")
ruta_grande = Path("materiales/U1_catalogo_grande.csv")

df_pequeno = pd.read_csv(ruta_pequeno)
df_grande = pd.read_csv(ruta_grande)

df_pequeno.head()
```

## 3. Comparar el volumen

```python
df_pequeno.shape, df_grande.shape
```

```python
memoria_pequeno = df_pequeno.memory_usage(deep=True).sum() / 1024
memoria_grande = df_grande.memory_usage(deep=True).sum() / 1024 / 1024

print(f"Catálogo pequeño: {memoria_pequeno:.1f} KB")
print(f"Catálogo grande: {memoria_grande:.1f} MB")
```

## 4. Comparar la velocidad

```python
def contar_documentos_largos(df):
    contador = 0
    for _, fila in df.iterrows():
        if fila["num_palabras"] > 500:
            contador += 1
    return contador
```

```python
inicio = time.perf_counter()
largos_pequeno = contar_documentos_largos(df_pequeno)
duracion_pequeno = time.perf_counter() - inicio

inicio = time.perf_counter()
largos_grande = contar_documentos_largos(df_grande)
duracion_grande = time.perf_counter() - inicio

print(f"Catálogo pequeño: {largos_pequeno} documentos largos, {duracion_pequeno * 1000:.2f} ms")
print(f"Catálogo grande: {largos_grande} documentos largos, {duracion_grande * 1000:.2f} ms")
```

## 5. Comparar la variedad

```python
for columna in ["emisora", "idioma", "tipo_contenido"]:
    print(columna, "->", df_pequeno[columna].nunique(), "vs", df_grande[columna].nunique())
```
